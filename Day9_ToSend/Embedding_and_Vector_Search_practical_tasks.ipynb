After loading the model, call the 'encode()' method on the model object to create a vector representation of a specific text sentence. Use your own text string  as the parameter.

```python
sentence = "There are many pre-trained embedding models available on Hugging Face that you can use to create vector embeddings."  #your sentence
embedding = model.encode(sentence)
embedding
```

Create vector representations for several text sentences. Place the text strings in a list and use this list as an argument. Use 8-10 sentences of 20-25 words each.  Call the 'encode()' method on the model object with the list of sentences as an argument.

```python
sentences_list = ["Vilnius Old Town is full of baroque churches, narrow cobblestone streets and small cafes that attract thousands of tourists every single summer.",
    "Trakai Castle stands on an island in Lake Galvė, and visitors cross a long wooden bridge to reach its red brick towers.",
    "Basketball is the most popular sport in Lithuania, and people gather in bars and homes to watch every national team game together.",
    "Many Lithuanian children start playing basketball at school, dreaming of one day joining a professional team and playing in big European arenas.",
    "Cepelinai are large potato dumplings filled with minced meat, usually served hot with sour cream and crispy fried bacon on top.",
    "Šaltibarščiai is a cold pink beetroot soup made with kefir, cucumbers and dill, and it is served with warm boiled potatoes.",
    "The Curonian Spit is a long, narrow strip of sand dunes and pine forests that separates the Baltic Sea from the lagoon.",
    "Palanga is a popular seaside resort where families walk along the pier at sunset and spend long summer days on the beach.",
    "Learning Python helps students analyse data, automate boring tasks and build small programs that solve real problems in their daily work.",
    "Vector databases store numeric embeddings of text, so applications can quickly find documents with similar meaning instead of matching exact words.",]

embeddings = model.encode(sentences_list)
embeddings
```

Write and test a function named 'cos_similarity_calculation' that determines the semantic similarity between the sentences in your list and any text sentence using their vector representations and the cosine distance as a similarity measure.  

```python
def cos_similarity_calculation(query, sentences):
    query_embedding = model.encode(query)
    sentence_embeddings = model.encode(sentences)
    return util.cos_sim(query_embedding, sentence_embeddings)[0]

query = "What traditional food should I try?"
scores = cos_similarity_calculation(query, sentences_list)

for sentence, score in zip(sentences_list, scores):
    print(f"{score:.4f}  {sentence}")
```

Create a function that determines the cosine similarity between a vector and a batch of vectors using the cosine distance formula and the numpy library. Add code to demonstrate how to use this function.

```python
import numpy as np

def cosine_similarity_batch(vector, batch):
    return (batch @ vector) / (np.linalg.norm(batch, axis=1) * np.linalg.norm(vector))

query = "Where can I relax by the sea?"
query_vector = model.encode(query)
scores = cosine_similarity_batch(query_vector, embeddings)

for sentence, score in zip(sentences_list, scores):
    print(f"{score:.4f}  {sentence}")
```
